## **1. Desempeno mensual.**
---

```python
# Query
sql_query = '''
              SELECT
              strftime('%Y-%m', order_date) AS MES,
              SUM(quantity * unit_price * (1 - discount_pct)) AS INGRESO_NETO,
              COUNT (DISTINCT o.order_id) AS ORDENES_COMPLETADAS,
              COUNT (DISTINCT customer_id) AS CLIENTES_UNICOS,
              (SUM(quantity * unit_price * (1 - discount_pct)) / COUNT(DISTINCT o.order_id)) AS TICKET_PROMEDIO
              FROM order_items oi
              INNER JOIN orders o
              ON oi.order_id = o.order_id
              WHERE 1=1
              AND status='Completada'
              GROUP BY MES
              '''
# Ejecuta el query
result_df = pandasql_query(sql_query)
# Muestra el resultado
display(result_df)
```

## **2. Categoria lider.**
---

```python
# Query
sql_query = '''
              WITH INGRESOS_CATEGORIA AS(
              SELECT
              strftime('%Y-%m', order_date) AS MES,
              SUM(quantity * oi.unit_price * (1 - discount_pct)) AS INGRESO_CATEGORIA,
              category CATEGORIA
              FROM order_items oi
              INNER JOIN orders o
              ON oi.order_id = o.order_id
              INNER JOIN products p
              ON oi.product_id = p.product_id
              WHERE 1=1
              AND status='Completada'
              GROUP BY MES, CATEGORIA
              ORDER BY MES DESC
              ),

              INGRESO_TOTAL AS (
              SELECT
              MES,
              CATEGORIA,
              INGRESO_CATEGORIA,
              SUM(INGRESO_CATEGORIA) OVER (
              PARTITION BY MES
              ) AS INGRESO_TOTAL_MES,
              INGRESO_CATEGORIA  /
              SUM(INGRESO_CATEGORIA) OVER (
              PARTITION BY MES
              )AS PORCENTAJE_MES
              FROM INGRESOS_CATEGORIA
              ),

              MAYOR_INGRESO AS (
              SELECT
              MES,
              CATEGORIA,
              INGRESO_CATEGORIA,
              PORCENTAJE_MES,
              ROW_NUMBER() OVER (
              PARTITION BY MES
              ORDER BY INGRESO_CATEGORIA DESC
              ) AS RN
              FROM INGRESO_TOTAL
              )


              SELECT
              MES,
              CATEGORIA,
              INGRESO_CATEGORIA,
              PORCENTAJE_MES
              FROM MAYOR_INGRESO
              WHERE 1=1
              AND RN = 1
              ORDER BY MES DESC
              '''
# Ejecuta el query
result_df = pandasql_query(sql_query)
# Muestra el resultado
display(result_df)
```

## **3. Clientes recurrentes.**
---

```python
#QUERY
sql_query = '''
              WITH ORDENES_CLIENTE AS (
              SELECT
              o.customer_id,
              COUNT(DISTINCT o.order_id) AS NUM_ORDENES
              FROM orders o
              WHERE 1=1
              AND status='Completada'
              GROUP BY o.customer_id
              ),

              CLIENTES_REGION AS (
              SELECT
              c.region,
              c.customer_id,
              oc.NUM_ORDENES
              FROM customers c
              INNER JOIN ORDENES_CLIENTE oc
              ON c.customer_id = oc.customer_id
              )

              SELECT
              region AS REGION,
              COUNT(*) AS TOTAL_CLIENTES_COMPRADORES,
              SUM(
              CASE
              WHEN NUM_ORDENES >= 2 THEN 1
              ELSE 0
              END
              ) AS CLIENTES_2_O_MAS_ORDENES,
              ROUND(
              100.0 * SUM(
              CASE
              WHEN NUM_ORDENES >= 2 THEN 1
              ELSE 0
              END
              ) / COUNT(*),
              2
              ) AS PORCENTAJE_CLIENTES_RECURRENTES
              FROM CLIENTES_REGION
              GROUP BY REGION
              ORDER BY PORCENTAJE_CLIENTES_RECURRENTES DESC
              '''
# Ejecuta el query
result_df = pandasql_query(sql_query)
# Muestra el resultado
display(result_df)
```

## **4. Riesgo operativo.**
---

```python
# Query
sql_query = '''
              SELECT
              CAST(SUM (CASE WHEN o.status= 'Cancelada' THEN 1 ELSE 0 END) AS REAL)  / COUNT(DISTINCT o.order_id) AS TASA_CANCELACION,
              strftime('%Y-%m', order_date) AS MES,
              o.channel AS CANAL,
              COUNT(DISTINCT o.order_id) AS TOTAL_ORDENES
              FROM orders o
              WHERE 1=1
              GROUP BY MES,CANAL
              HAVING TASA_CANCELACION>=.15
              '''
# Ejecuta el query
result_df = pandasql_query(sql_query)
# Muestra el resultado
display(result_df)
```

## **2. Tabla Semanal.**
---

```python
df = pd.merge(order_items, orders, on = 'order_id', suffixes = [' itemes', ''])
```

```python
df['order_date'] = pd.to_datetime(df['order_date'])
```

```python
df['ingreso_neto'] = (df['quantity'] * df['unit_price']) * (1 - df['discount_pct'])
```

```python
df = df[df['status'] == "Completada"].copy()
df.head()
```

```python
df_semanal = df.groupby(pd.Grouper(key = 'order_date', freq =  'W')).agg(ingreso_neto = ('ingreso_neto', 'sum'), ordenes_completadas = ('order_id', 'nunique')).reset_index()
df_semanal['ticket_promedio'] = df_semanal['ingreso_neto'] / df_semanal['ordenes_completadas']
df_semanal
```

## **3. Semanas con disminucion del ingreso neto.**
---

```python
df_semanal['ingreso_neto anterior'] = df_semanal['ingreso_neto'].shift(1)
```

```python
df_semanal['Disminucion'] = np.where(df_semanal['ingreso_neto'] < df_semanal['ingreso_neto anterior'] * 0.85, 1, 0)
```

```python
df_semanal['% Disminucion'] = (df_semanal['ingreso_neto'] / df_semanal['ingreso_neto anterior']) - 1
```

```python
df_semanal
```

```python
plt.figure(figsize=(12, 6))
sns.lineplot(data=df_semanal, x='order_date', y='ingreso_neto', color='red')
plt.title('Tendencia Semanal de Ingreso Neto',
          fontsize = 20,
          color = 'red',fontweight='bold')

plt.xlabel('Fecha Orden',fontsize = 20,
           color = 'red',fontweight='bold')

plt.ylabel('Ingresoso Netos',fontsize = 20,
           color = 'red',fontweight='bold')

plt.xticks(fontsize=14, color='red',fontweight='bold')
plt.yticks(fontsize=14, color='red',fontweight='bold')
plt.grid(True)
plt.show()
```

## **4) Segmentación util de clientes**
---

Se busca segmentar a los clientes por patrones de las órdenes canceladas y devueltas, analizaremos su distribución por región de cliente y por producto.

```python
# Ordenes canceladas o devueltas
segmentacion = orders[orders['status'].isin(['Cancelada', 'Devuelta'])]

# Segmentación por Región
region = pd.merge(segmentacion, customers, on='customer_id', how='inner')
print("\nÓrdenes Canceladas/Devueltas por Región:")
display(region['region'].value_counts())

# Segmentación por Producto
producto = pd.merge(segmentacion, order_items, on='order_id', how='inner')
cancelled_returned_products_details = pd.merge(producto, products, on='product_id', how='inner')
print("\nTop 5 Productos más Cancelados/Devueltos:")
display(cancelled_returned_products_details['product_name'].value_counts().head(5))
```

### **Uso de Negocio Simple:**

Identificar qué regiones tienen más cancelaciones/devoluciones puede indicar problemas de logística locales o expectativas de cliente diferentes. Saber qué productos se devuelven más puede señalar problemas de calidad, descripciones poco claras o que el producto no satisface la necesidad real del cliente. Con esta información, la empresa puede enfocar sus esfuerzos en mejorar procesos o productos específicos para reducir estas incidencias.
